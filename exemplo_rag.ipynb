```python
from langchain_core.documents import Document

docs = [
    Document(page_content="A politica de férias da empresa permite 30 dias por ano"),
    Document(page_content="O time de engenharia usa LangChain para construir agentes de IA"),
    Document(page_content="Os benefícios incluiem plano de saude e vale alimentação"),
    Document(page_content="O expediente padrão é de segunda a sexta, das 9h as 18h"),

]
```

```python
from langchain_openai import OpenAIEmbeddings
from langchain_core.vectorstores import InMemoryVectorStore
from dotenv import load_dotenv

load_dotenv()

embeddings = OpenAIEmbeddings(model="text-embedding-3-large")

vectorstore = InMemoryVectorStore.from_documents(
    documents=docs,
    embedding=embeddings
)
```

```python
retriver = vectorstore.as_retriever(search_kwargs={"k":2})
```

```python
perguntas = [
    "Quantos dias de férias eu tenho no ano?",
    "Quais os beneficios que a empresa oferece?",
    "Qual ferramenta a equipe usa?",
    "Qual o expediente padrão?"
]
```

```python
for pergunta in perguntas:
    print("=" * 60)
    print("PERGUNTA: ", pergunta)
    resultados = retriver.invoke(pergunta)
    for i, doc in enumerate(resultados, start=1):
        print(f"Resultado {i}: ", doc.page_content)
```
